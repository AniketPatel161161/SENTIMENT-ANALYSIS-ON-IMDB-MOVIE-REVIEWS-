# Q-Learning for CliffWalking Environment

This notebook demonstrates the implementation of a Q-learning agent to solve the CliffWalking-v1 environment from Gymnasium. The goal is to train an agent to navigate from a starting point to a goal, avoiding cliffs, and maximizing cumulative reward.

```python
import gymnasium as gym
import numpy as np
import random
```

## 1. Environment Setup and Library Imports

We import the necessary libraries: `gymnasium` for the environment, `numpy` for numerical operations (especially for the Q-table), and `random` for implementing the epsilon-greedy exploration strategy.

```python
gamma = 0.99
epsilon = 0.1
alpha = 0.5
episodes = 500
```

## 2. Hyperparameters

These are the key parameters that control the learning process of the Q-learning algorithm:

*   `gamma` (Discount Factor): Determines the importance of future rewards. A higher value (closer to 1) makes the agent consider future rewards more heavily.
*   `epsilon` (Exploration Rate): Controls the trade-off between exploration (trying new actions) and exploitation (choosing the best-known action). A higher epsilon means more exploration.
*   `alpha` (Learning Rate): Determines how much the new Q-value overwrites the old Q-value. A higher alpha means faster learning, but can lead to instability.
*   `episodes`: The total number of training iterations (episodes) the agent will run through the environment.

## 3. Q-Table Initialization

The Q-table is the core component of Q-learning. It's a matrix where rows represent states and columns represent actions. Each cell `Q[s, a]` stores the estimated maximum future reward for taking action `a` in state `s`.

*   **Dimensions**: The CliffWalking environment has 48 possible states (12 columns x 4 rows) and 4 possible actions (up, down, left, right). So, the Q-table is initialized as a 48x4 NumPy array with all zeros.

```python
Q = np.zeros((48, 4))
```

```python
# epsilon-policy
def epsilon_greedy(state):
    if random.random() < epsilon:
        return env.action_space.sample() # explore
    else:
        return np.argmax(Q[state])
```

## 4. Epsilon-Greedy Policy

This function defines how the agent chooses an action during training. It balances exploration and exploitation:

*   **Exploration**: With a probability of `epsilon`, the agent chooses a random action from the environment's action space. This helps the agent discover new paths and potentially better rewards.
*   **Exploitation**: With a probability of `1 - epsilon`, the agent chooses the action that has the highest Q-value for the current state. This means it selects the action it currently believes is the best based on its learned experience.

## Q Learning

```python
for episode in range(episodes):
    render = (episode % 50 == 0)
    if render :
        env = gym.make("CliffWalking-v1" , render_mode = "human")
    else:
        env= gym.make("CliffWalking-v1")
    done = False
    state,_ = env.reset()
    episode_len = 0
    tot_reward = 0
    while not done:
        action = epsilon_greedy(state)
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated
        # Q-learning update
        Q[state, action] += alpha * (reward + gamma * np.max(Q[next_state]) - Q[state, action])
        state = next_state
        episode_len += 1
        tot_reward += reward
    print(f"episode={episode+1}/500 & total reward = {tot_reward} & episode len = {episode_len}")
    env.close()
```

## 5. Q-Learning Training Loop

This is the main training phase where the agent interacts with the environment and updates its Q-table.

*   **Episode Iteration**: The loop runs for a predefined number of `episodes`.
*   **Environment Initialization**: For every 50th episode, the environment is rendered in `human` mode to visualize the agent's progress; otherwise, it runs without rendering.
*   **State and Reward Tracking**: For each step within an episode, the agent selects an `action` using the `epsilon_greedy` policy, observes the `next_state` and `reward` from the environment, and checks if the episode `done` (terminated or truncated).
*   **Q-Value Update (Bellman Equation)**: The core of Q-learning is this update rule:
    `Q[state, action] += alpha * (reward + gamma * np.max(Q[next_state]) - Q[state, action])`
    This formula updates the Q-value for the current state-action pair based on the immediate `reward`, the `gamma`-discounted maximum Q-value of the `next_state`, and the current Q-value of the `state-action` pair.
*   **Progress Tracking**: The total reward and episode length are printed for each episode to monitor the training progress.

```python
# what did our agent learn?
env = gym.make("CliffWalking-v1", render_mode="human")
state, _ = env.reset()
total_reward = 0
epsiode_len = 0
done = False
while not done:
    action = np.argmax(Q[state])
    state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated
    epsiode_len += 1
    total_reward += reward
print(f"total reward = {total_reward} & epsiode len = {epsiode_len}")
env.close()
```

## 6. Evaluating the Learned Policy

After training, this section evaluates the agent's learned policy. Unlike the training phase, during evaluation, the agent acts purely greedily, meaning it always chooses the action with the highest Q-value for any given state (no exploration).

*   **Greedy Action Selection**: `action = np.argmax(Q[state])` selects the action that maximizes the Q-value for the current state.
*   **Performance Metrics**: The total reward and episode length are recorded and printed to assess how well the agent performs after learning, without any random exploration.

## 7. How are we selecting the best move according to the Q-values?

As explained in the 'Epsilon-Greedy Policy' and 'Evaluating the Learned Policy' sections, the selection of the 'best' move depends on the phase:

*   **During Training (Epsilon-Greedy Policy)**: The agent mostly exploits its current knowledge by choosing `np.argmax(Q[state])`, which returns the action index corresponding to the highest Q-value for the current state. However, it also explores randomly (`env.action_space.sample()`) with a probability of `epsilon`.

*   **During Evaluation (Greedy Policy)**: Once the agent is trained, we want to see its optimal performance. In this phase, `epsilon` is effectively 0, and the agent *always* chooses the action with the highest Q-value for the current state using `np.argmax(Q[state])`. This is pure exploitation of the learned Q-table.
